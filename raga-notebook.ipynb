## Raga Parser

```python
from raga.parser import SwarlipiExtractor
```

```python
default_dataset = "datasets/ragas"
extractor = SwarlipiExtractor(default_dataset)

titles = extractor.find_compositions_by_arohana("s-R-g")
for title in titles[:5]:  # Show first 5
    print(f"  - {title}")
if len(titles) > 5:
    print(f"  ... and {len(titles) - 5} more")
```

```python
raags = extractor.get_all_raags()
for raag in raags:
    print(f"  - {raag}")
```

```python
freq_dist = extractor.get_note_frequency_distribution()
if freq_dist:
    sample = freq_dist[0]
    print(f"  Composition: {sample['title']}")
    print(f"  Raag: {sample['raag']}")
    print(f"  Note counts: {sample['note_counts']}")
```

### Pitch Extractor

```python
from raga.pitch_extractor import PitchExtractor, plot_pitch_curve

xml_file = default_dataset + "/Bhairav-1-168.xml"

tonic_freq = 261.63
extractor = PitchExtractor(tonic_freq=tonic_freq)
curve = extractor.extract_pitch_curve(xml_file)
plot_pitch_curve(curve, show_note_lines=True)
```
